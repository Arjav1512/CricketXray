```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batting = pd.read_csv("/content/t20.csv")
bowling = pd.read_csv("/content/Bowling_t20.csv")
fielding = pd.read_csv("/content/Fielding_t20.csv")
```

```python
print("BATTING")
print(batting.shape)
print(batting.columns.tolist())

print("\nBOWLING")
print(bowling.shape)
print(bowling.columns.tolist())

print("\nFIELDING")
print(fielding.shape)
print(fielding.columns.tolist())
```

```python
batting.head()
bowling.head()
fielding.head()
```

```python
print("Batting = Bowling:", batting['Player'].equals(bowling['Player']))
print("Batting = Fielding:", batting['Player'].equals(fielding['Player']))
```

```python
print("BATTING:")
print(batting['Player'].head(10).tolist())

print("\nBOWLING:")
print(bowling['Player'].head(10).tolist())

print("\nFIELDING:")
print(fielding['Player'].head(10).tolist())
```

```python
bat_players = set(batting['Player'])
bowl_players = set(bowling['Player'])
field_players = set(fielding['Player'])

print("Batting players:", len(bat_players))
print("Bowling players:", len(bowl_players))
print("Fielding players:", len(field_players))

print("Common in all 3:", len(bat_players & bowl_players & field_players))
```

```python
print("In batting but not bowling:")
print(list(bat_players - bowl_players)[:10])

print("\nIn batting but not fielding:")
print(list(bat_players - field_players)[:10])
```

```python
batting_clean = batting.drop(
    columns=[col for col in batting.columns if col.startswith("Unnamed")]
)

bowling_clean = bowling.drop(
    columns=[col for col in bowling.columns if col.startswith("Unnamed")]
)

fielding_clean = fielding.drop(
    columns=[col for col in fielding.columns if col.startswith("Unnamed")]
)

print(batting_clean.columns)
print(bowling_clean.columns)
print(fielding_clean.columns)
```

```python
print("Batting duplicates:", batting_clean['Player'].duplicated().sum())
print("Bowling duplicates:", bowling_clean['Player'].duplicated().sum())
print("Fielding duplicates:", fielding_clean['Player'].duplicated().sum())
```

```python
combined = batting_clean.merge(
    bowling_clean,
    on="Player",
    how="inner",
    suffixes=("_Bat", "_Bowl")
)

combined = combined.merge(
    fielding_clean,
    on="Player",
    how="inner",
    suffixes=("", "_Field")
)
```
